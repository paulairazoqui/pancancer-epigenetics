# 451 — Locus-Level Methylation–Expression Characterization

## Objective

Characterize locus-level DNA-methylation–gene-expression relationships associated
with the frozen tumor-side methylation context of the three Phase 4 cross-system
consensus transcriptomic programs in the TCGA primary-tumor cohort.

The analysis integrates frozen TCGA DNA-methylation and RNA-seq representations
from Phase 2 with the frozen tumor methylation-component context preserved through
Phase 4.

The objective is to determine whether strongly contributing CpG loci from the
relevant frozen methylation components show reproducible associations with
expression of defensibly annotated genes while preserving lineage structure,
platform boundaries, upstream limitations, and explicit separation between
statistical association and biological mechanism.

Inverse promoter-associated methylation–expression relationships are interpreted
as compatible with an epigenetic regulatory hypothesis. They do not establish
methylation-mediated transcriptional repression or biological causality.

## Frozen upstream inputs

Notebook 451 operates on frozen upstream objects:

- the TCGA primary-tumor multi-omic cohort and exact case-level sample mapping
  from notebook 203;
- the frozen TMM-logCPM RNA-seq representation generated in notebook 205;
- the frozen HM450, HM27, and shared HM27/HM450 DNA-methylation
  representations from notebook 203;
- the frozen methylation probe mapping from notebook 203;
- the frozen methylation ICA probe loadings from notebook 205;
- the three frozen Phase 4 consensus transcriptomic programs and their
  tumor-side methylation-component context; and
- frozen tumor-side confounder covariates from notebook 204.

Upstream cohort membership, sample identity, RNA-seq normalization, methylation
QC, methylation-component identities, component orientations, consensus-program
identities, and program definitions are not recomputed or modified on the basis
of locus-level results.

## Frozen program-context boundary

Notebook 451 is restricted to the tumor-side methylation components already
linked to the three frozen consensus transcriptomic programs:

- `CONSENSUS_TX_01` — `METH_IC128`;
- `CONSENSUS_TX_02` — `METH_IC050`; and
- `CONSENSUS_TX_03` — `METH_IC169` and `METH_IC128`.

Because `METH_IC128` occurs in more than one consensus-program context, statistical
inference is performed on unique CpG–gene units rather than duplicated by program
membership.

Consensus-program and methylation-component memberships are preserved as
downstream contextual annotations.

No locus-level result may rescue, exclude, reweight, rename, reorient, or redefine
a frozen consensus program or its upstream tumor methylation context.

## Locus and annotation boundary

The primary locus universe is defined from the strongest signed loading tails of
the relevant frozen methylation components before methylation–expression
association results are inspected.

For each unique relevant methylation component, the primary locus rule retains:

- the 50 most negative-loading CpGs; and
- the 50 most positive-loading CpGs.

This reuses the pre-existing loading-tail convention established upstream rather
than selecting a new threshold from notebook-451 association results.

CpGs shared across methylation components are retained once as loci while
preserving every applicable component and consensus-program membership.

Biological CpG-to-gene and regulatory-context annotation must come from a
versioned, provenance-tracked, frozen annotation authority before inferential
testing begins.

CpG-to-gene mappings are not selected, rescued, or reassigned according to
observed expression associations, cancer relevance, pathway membership,
druggability, or expected biological interpretation.

## Platform and expression boundary

HM450 is the primary methylation platform for locus-level inference because the
relevant methylation-component discovery was performed in the frozen HM450
analysis space and HM450 provides the larger platform-specific tumor cohort.

HM27 is treated as a prespecified platform-sensitivity layer where locus and
lineage coverage permit evaluation.

HM27 and HM450 samples are not naively pooled into a single primary regression
model.

Gene expression is taken directly from the frozen Phase 2 TMM-logCPM
representation. Notebook 451 does not refit expression filtering or
normalization.

Primary methylation inference uses the M-value transformation convention already
established upstream. Beta values remain available for descriptive interpretation.

## Lineage-aware analytical boundary

Cancer lineage is treated as a required structural component of the analysis.

The primary statistical unit is the unique annotated:

`CpG × gene`

pair.

Primary CpG–gene models are restricted to TCGA projects satisfying prespecified
complete-case and within-project variation requirements.

The primary model is:

`gene_expression ~ CpG_methylation + C(project_id)`

with heteroskedasticity-consistent HC3 uncertainty.

Within-project estimates and leave-one-project-out analyses characterize
directional consistency, lineage dependence, and sensitivity to individual
projects for primary FDR-supported associations.

A pooled association alone is not sufficient to establish cross-project
recurrence.

Cross-project recurrence additionally requires prespecified directional support
across eligible projects and leave-one-project-out stability.

Lineage-aware recurrence describes reproducibility of an association across
eligible TCGA projects. It does not establish biological universality across
cancer types.

## Confounding, multiplicity, and sensitivity analyses

The primary model adjusts for TCGA project while preserving the frozen primary
estimand and maximum eligible HM450 cohort.

Prespecified sensitivity analyses evaluate, where applicable:

- tumor purity;
- proliferation;
- leukocyte fraction for loci belonging to `METH_IC128`; and
- HM27 platform consistency.

The additional cellular-composition sensitivity for `METH_IC128` is motivated by
a frozen upstream interpretive limitation and is not activated according to
notebook-451 results.

Sensitivity analyses characterize stability of primary associations and cannot
rescue associations that fail the primary inferential framework.

Multiplicity is controlled using Benjamini–Hochberg FDR across the complete
frozen unique CpG–gene primary hypothesis family.

FDR families are not subdivided post hoc by consensus program, methylation
component, regulatory annotation, association direction, gene class, or TCGA
project.

## Interpretation boundary

A negative methylation coefficient indicates that higher methylation at the CpG
is associated with lower expression of the annotated gene.

When such a relationship occurs at a promoter-associated locus it may be
described as an inverse promoter-associated methylation–expression relationship
compatible with an epigenetic regulatory hypothesis.

It does not establish:

- promoter silencing;
- methylation-mediated repression;
- a causal regulatory interaction;
- a mechanism of drug resistance;
- a validated biomarker;
- a validated therapeutic target; or
- therapeutic reversibility.

Positive, inverse, heterogeneous, lineage-restricted, technically limited, or
null results are all valid scientific outcomes.

## Scope and methodological boundary

Notebook 451 does not:

- redefine or reoptimize consensus programs;
- refit methylation or transcriptomic latent components;
- assign resistant or sensitive labels to TCGA tumors;
- select loci or genes using pharmacogenomic phenotype;
- infer causal regulation from methylation–expression association;
- choose CpG-to-gene mappings according to association results;
- pool HM27 and HM450 naïvely;
- treat HM27 as independent external biological validation;
- use sensitivity analyses to redefine primary significance;
- promote individual genes as validated targets; or
- use downstream functional, pharmacogenomic, perturbational, or therapeutic
  evidence to select locus-level associations.

The resulting CpG–gene relationships constitute secondary locus-level molecular
context for the frozen consensus programs.

## Expected outputs

Notebook 451 will produce reproducible locus-level characterization under:

`data/processed/secondary_characterization/`

The analysis will preserve:

- the deterministic strong-loading CpG universe and CpG–gene annotation
  eligibility;
- the complete frozen primary CpG–gene association family;
- lineage-aware recurrence characterization;
- prespecified sensitivity evidence;
- promoter/regulatory-context annotation;
- consensus-program and methylation-component context; and
- explicit negative, heterogeneous, and technically limited results.

Only downstream-consumable artifacts justified by the completed analysis will be
persisted and registered. Intermediate diagnostic objects remain reproducible
within the notebook unless separate persistence is required for downstream
provenance or reuse.

```python
# =============================================================================
# Define frozen notebook-451 methylation-component context
# =============================================================================

methylation_component_context = (
    consensus_tumor_arm_context[
        [
            "consensus_program_id",
            "candidate_pair",
            "rna_component",
            "methylation_component",
            "final_audit_status",
            "scientific_priority",
            "primary_limitation",
            "consensus_handling",
        ]
    ]
    .drop_duplicates()
    .sort_values(
        ["consensus_program_id", "methylation_component", "candidate_pair"]
    )
    .reset_index(drop=True)
)

relevant_methylation_components = tuple(
    sorted(
        methylation_component_context[
            "methylation_component"
        ].dropna().unique()
    )
)

missing_loading_components = sorted(
    set(relevant_methylation_components)
    - set(methylation_probe_loadings.columns)
)

print(
    "Consensus programs represented:",
    methylation_component_context["consensus_program_id"].nunique(),
)
print(
    "Tumor methylation arms represented:",
    methylation_component_context.shape[0],
)
print(
    "Unique methylation components:",
    relevant_methylation_components,
)
print(
    "All relevant components present in frozen loadings:",
    not missing_loading_components,
)

methylation_component_context
```

```python
# =============================================================================
# Construct strong-loading CpG membership universe
# =============================================================================

LOADING_TAIL_SIZE = 50

strong_loading_memberships = []

for component in relevant_methylation_components:
    component_loadings = methylation_probe_loadings[
        ["model_matrix_row_index", "probe_id", component]
    ].rename(columns={component: "loading"})

    negative_tail = (
        component_loadings
        .sort_values(
            ["loading", "probe_id"],
            ascending=[True, True],
        )
        .head(LOADING_TAIL_SIZE)
        .assign(
            methylation_component=component,
            loading_tail="negative",
        )
    )

    positive_tail = (
        component_loadings
        .sort_values(
            ["loading", "probe_id"],
            ascending=[False, True],
        )
        .head(LOADING_TAIL_SIZE)
        .assign(
            methylation_component=component,
            loading_tail="positive",
        )
    )

    strong_loading_memberships.extend(
        [negative_tail, positive_tail]
    )

strong_loading_cpg_memberships = (
    pd.concat(
        strong_loading_memberships,
        ignore_index=True,
    )
    [
        [
            "probe_id",
            "model_matrix_row_index",
            "methylation_component",
            "loading",
            "loading_tail",
        ]
    ]
    .sort_values(
        ["methylation_component", "loading_tail", "loading"],
        ascending=[True, True, True],
    )
    .reset_index(drop=True)
)

expected_memberships = (
    len(relevant_methylation_components)
    * 2
    * LOADING_TAIL_SIZE
)

print("Expected component-tail memberships:", expected_memberships)
print(
    "Observed component-tail memberships:",
    strong_loading_cpg_memberships.shape[0],
)
print(
    "Unique strong-loading CpGs:",
    strong_loading_cpg_memberships["probe_id"].nunique(),
)
print(
    "Duplicate membership rows:",
    strong_loading_cpg_memberships.duplicated(
        ["probe_id", "methylation_component", "loading_tail"]
    ).sum(),
)

strong_loading_cpg_memberships.groupby(
    ["methylation_component", "loading_tail"]
).size()
```

```python
# =============================================================================
# Construct unique strong-loading CpG locus universe
# =============================================================================

probe_index_counts = (
    strong_loading_cpg_memberships
    .groupby("probe_id")["model_matrix_row_index"]
    .nunique()
)

if not probe_index_counts.eq(1).all():
    raise ValueError(
        "At least one strong-loading CpG maps to multiple model matrix rows."
    )

strong_loading_cpg_universe = (
    strong_loading_cpg_memberships
    .groupby(
        ["probe_id", "model_matrix_row_index"],
        as_index=False,
    )
    .agg(
        n_methylation_components=(
            "methylation_component",
            "nunique",
        ),
        methylation_components=(
            "methylation_component",
            lambda values: "|".join(sorted(set(values))),
        ),
        component_tail_memberships=(
            "methylation_component",
            lambda values: "|".join(
                sorted(set(values))
            ),
        ),
    )
)

component_tail_context = (
    strong_loading_cpg_memberships
    .assign(
        component_tail=lambda frame: (
            frame["methylation_component"]
            + ":"
            + frame["loading_tail"]
        )
    )
    .groupby("probe_id")["component_tail"]
    .agg(lambda values: "|".join(sorted(set(values))))
)

strong_loading_cpg_universe[
    "component_tail_memberships"
] = strong_loading_cpg_universe["probe_id"].map(
    component_tail_context
)

strong_loading_cpg_universe = (
    strong_loading_cpg_universe
    .sort_values("probe_id")
    .reset_index(drop=True)
)

print(
    "Unique strong-loading CpGs:",
    strong_loading_cpg_universe.shape[0],
)
print(
    "CpGs represented in >1 methylation component:",
    int(
        strong_loading_cpg_universe[
            "n_methylation_components"
        ].gt(1).sum()
    ),
)
print(
    "Maximum component memberships for one CpG:",
    int(
        strong_loading_cpg_universe[
            "n_methylation_components"
        ].max()
    ),
)

strong_loading_cpg_universe[
    "n_methylation_components"
].value_counts().sort_index()
```

```python
# =============================================================================
# Map strong-loading CpGs to frozen HM450 matrix rows
# =============================================================================

hm450_probe_mapping = (
    probe_mapping.loc[
        probe_mapping["representation"].eq("hm450"),
        [
            "probe_id",
            "matrix_row_index",
            "probe_missing_beta_fraction",
            "probe_qc_eligible",
            "hm450_missing_beta_fraction",
            "hm450_probe_qc_eligible",
        ],
    ]
    .rename(
        columns={
            "matrix_row_index": "hm450_matrix_row_index",
        }
    )
)

hm450_strong_probe_mapping = (
    strong_loading_cpg_universe
    .merge(
        hm450_probe_mapping,
        on="probe_id",
        how="left",
        validate="one_to_one",
    )
    .sort_values("probe_id")
    .reset_index(drop=True)
)

if hm450_strong_probe_mapping["hm450_matrix_row_index"].isna().any():
    raise ValueError(
        "At least one strong-loading CpG lacks an HM450 matrix-row mapping."
    )

print(
    "Strong-loading CpGs:",
    hm450_strong_probe_mapping.shape[0],
)
print(
    "Mapped to HM450 rows:",
    hm450_strong_probe_mapping[
        "hm450_matrix_row_index"
    ].notna().sum(),
)
print(
    "HM450 QC-eligible:",
    hm450_strong_probe_mapping[
        "hm450_probe_qc_eligible"
    ].fillna(False).sum(),
)
print(
    "Unique HM450 matrix rows:",
    hm450_strong_probe_mapping[
        "hm450_matrix_row_index"
    ].nunique(),
)
```

### **CpG biological-annotation checkpoint**

The strong-loading CpG universe and the frozen RNA expression gene axis have now
been defined independently of methylation–expression association results.

Before constructing CpG–gene pairs, notebook 451 requires one authoritative,
versioned biological annotation resource for HM450 probes.

The annotation resource must provide, where available:

- CpG probe identity;
- genomic coordinates;
- genome build;
- gene association;
- promoter or other regulatory-context annotation; and
- sufficient provenance to reproduce the mapping deterministically.

The annotation authority must be frozen before any methylation–expression
association is calculated.

Primary CpG-to-gene assignment will follow the prespecified identifier policy:

1. exact Ensembl base-ID matching when the annotation resource provides a
   compatible stable Ensembl gene identifier;
2. otherwise exact gene-symbol matching only when the symbol is unambiguous in
   the frozen RNA expression matrix.

No fuzzy matching, nearest-gene rescue, manual alias substitution, or
result-driven reassignment is permitted in the primary analysis.

The frozen RNA expression representation contains 18,123 genes with unique
Ensembl base identifiers. Gene-symbol mapping is unambiguous for 18,121 genes.
`POLR2J3` is represented by two distinct Ensembl genes and therefore cannot be
assigned through symbol-only matching.

CpGs mapping to multiple defensible genes will retain every valid CpG–gene
relationship as separate candidate pairs. Repeated transcript-level annotations
of the same CpG–gene relationship will later be collapsed to one unique
inferential CpG–gene unit while preserving annotation metadata.

No methylation–expression inference proceeds until the annotation resource,
version, genome build, provenance, and local frozen representation have been
defined and audited.

### **Frozen CpG annotation authority**

Notebook 451 uses a single prespecified biological annotation authority for
HM450 CpG probes:

- annotation framework: `SeSAMe / sesameData`;
- methylation platform: `HM450`;
- genome build: `hg38`.

This decision is frozen before inspection of any locus-level
methylation–expression association.

The annotation resource will be used only to establish deterministic biological
context for the frozen strong-loading CpG universe, including, where available:

- genomic coordinates;
- CpG-to-gene relationships; and
- promoter or other regulatory-context annotation.

The selected annotation authority does not replace or modify the frozen
technical probe mapping produced in Phase 2. The Phase 2 probe mapping remains
authoritative for matrix-row identity, QC eligibility, platform membership, and
missingness.

Notebook 451 will therefore preserve two distinct annotation layers:

1. **technical probe identity and matrix mapping** inherited from Phase 2; and
2. **biological CpG annotation** derived from the frozen SeSAMe HM450 hg38
   resource.

No alternative annotation source may be substituted after association results
are inspected because it yields more favorable CpG-to-gene mappings or
regulatory interpretation.

The exact SeSAMe / sesameData version, resource identity, retrieval procedure,
local path, and cryptographic identity must be recorded before inferential
