# 451 — Locus-Level Methylation–Expression Characterization

## Objective

Characterize locus-level DNA-methylation–gene-expression relationships associated
with the frozen tumor-side methylation context of the three Phase 4 cross-system
consensus transcriptomic programs in the TCGA primary-tumor cohort.

The analysis integrates frozen TCGA DNA-methylation and RNA-seq representations
from Phase 2 with the frozen tumor methylation-component context preserved through
Phase 4.

The objective is to determine whether strongly contributing CpG loci from the
relevant frozen methylation components show reproducible associations with
expression of defensibly annotated genes while preserving lineage structure,
platform boundaries, upstream limitations, and explicit separation between
statistical association and biological mechanism.

Inverse promoter-associated methylation–expression relationships are interpreted
as compatible with an epigenetic regulatory hypothesis. They do not establish
methylation-mediated transcriptional repression or biological causality.

## Frozen upstream inputs

Notebook 451 operates on frozen upstream objects:

- the TCGA primary-tumor multi-omic cohort and exact case-level sample mapping
  from notebook 203;
- the frozen TMM-logCPM RNA-seq representation generated in notebook 205;
- the frozen HM450, HM27, and shared HM27/HM450 DNA-methylation
  representations from notebook 203;
- the frozen methylation probe mapping from notebook 203;
- the frozen methylation ICA probe loadings from notebook 205;
- the three frozen Phase 4 consensus transcriptomic programs and their
  tumor-side methylation-component context; and
- frozen tumor-side confounder covariates from notebook 204.

Upstream cohort membership, sample identity, RNA-seq normalization, methylation
QC, methylation-component identities, component orientations, consensus-program
identities, and program definitions are not recomputed or modified on the basis
of locus-level results.

## Frozen program-context boundary

Notebook 451 is restricted to the tumor-side methylation components already
linked to the three frozen consensus transcriptomic programs:

- `CONSENSUS_TX_01` — `METH_IC128`;
- `CONSENSUS_TX_02` — `METH_IC050`; and
- `CONSENSUS_TX_03` — `METH_IC169` and `METH_IC128`.

Because `METH_IC128` occurs in more than one consensus-program context, statistical
inference is performed on unique CpG–gene units rather than duplicated by program
membership.

Consensus-program and methylation-component memberships are preserved as
downstream contextual annotations.

No locus-level result may rescue, exclude, reweight, rename, reorient, or redefine
a frozen consensus program or its upstream tumor methylation context.

## Locus and annotation boundary

The primary locus universe is defined from the strongest signed loading tails of
the relevant frozen methylation components before methylation–expression
association results are inspected.

For each unique relevant methylation component, the primary locus rule retains:

- the 50 most negative-loading CpGs; and
- the 50 most positive-loading CpGs.

This reuses the pre-existing loading-tail convention established upstream rather
than selecting a new threshold from notebook-451 association results.

CpGs shared across methylation components are retained once as loci while
preserving every applicable component and consensus-program membership.

Biological CpG-to-gene and regulatory-context annotation must come from a
versioned, provenance-tracked, frozen annotation authority before inferential
testing begins.

CpG-to-gene mappings are not selected, rescued, or reassigned according to
observed expression associations, cancer relevance, pathway membership,
druggability, or expected biological interpretation.

## Platform and expression boundary

HM450 is the primary methylation platform for locus-level inference because the
relevant methylation-component discovery was performed in the frozen HM450
analysis space and HM450 provides the larger platform-specific tumor cohort.

HM27 is treated as a prespecified platform-sensitivity layer where locus and
lineage coverage permit evaluation.

HM27 and HM450 samples are not naively pooled into a single primary regression
model.

Gene expression is taken directly from the frozen Phase 2 TMM-logCPM
representation. Notebook 451 does not refit expression filtering or
normalization.

Primary methylation inference uses the M-value transformation convention already
established upstream. Beta values remain available for descriptive interpretation.

## Lineage-aware analytical boundary

Cancer lineage is treated as a required structural component of the analysis.

The primary statistical unit is the unique annotated:

`CpG × gene`

pair.

Primary CpG–gene models are restricted to TCGA projects satisfying prespecified
complete-case and within-project variation requirements.

The primary model is:

`gene_expression ~ CpG_methylation + C(project_id)`

with heteroskedasticity-consistent HC3 uncertainty.

Within-project estimates and leave-one-project-out analyses characterize
directional consistency, lineage dependence, and sensitivity to individual
projects for primary FDR-supported associations.

A pooled association alone is not sufficient to establish cross-project
recurrence.

Cross-project recurrence additionally requires prespecified directional support
across eligible projects and leave-one-project-out stability.

Lineage-aware recurrence describes reproducibility of an association across
eligible TCGA projects. It does not establish biological universality across
cancer types.

## Confounding, multiplicity, and sensitivity analyses

The primary model adjusts for TCGA project while preserving the frozen primary
estimand and maximum eligible HM450 cohort.

Prespecified sensitivity analyses evaluate, where applicable:

- tumor purity;
- proliferation;
- leukocyte fraction for loci belonging to `METH_IC128`; and
- HM27 platform consistency.

The additional cellular-composition sensitivity for `METH_IC128` is motivated by
a frozen upstream interpretive limitation and is not activated according to
notebook-451 results.

Sensitivity analyses characterize stability of primary associations and cannot
rescue associations that fail the primary inferential framework.

Multiplicity is controlled using Benjamini–Hochberg FDR across the complete
frozen unique CpG–gene primary hypothesis family.

FDR families are not subdivided post hoc by consensus program, methylation
component, regulatory annotation, association direction, gene class, or TCGA
project.

## Interpretation boundary

A negative methylation coefficient indicates that higher methylation at the CpG
is associated with lower expression of the annotated gene.

When such a relationship occurs at a promoter-associated locus it may be
described as an inverse promoter-associated methylation–expression relationship
compatible with an epigenetic regulatory hypothesis.

It does not establish:

- promoter silencing;
- methylation-mediated repression;
- a causal regulatory interaction;
- a mechanism of drug resistance;
- a validated biomarker;
- a validated therapeutic target; or
- therapeutic reversibility.

Positive, inverse, heterogeneous, lineage-restricted, technically limited, or
null results are all valid scientific outcomes.

## Scope and methodological boundary

Notebook 451 does not:

- redefine or reoptimize consensus programs;
- refit methylation or transcriptomic latent components;
- assign resistant or sensitive labels to TCGA tumors;
- select loci or genes using pharmacogenomic phenotype;
- infer causal regulation from methylation–expression association;
- choose CpG-to-gene mappings according to association results;
- pool HM27 and HM450 naïvely;
- treat HM27 as independent external biological validation;
- use sensitivity analyses to redefine primary significance;
- promote individual genes as validated targets; or
- use downstream functional, pharmacogenomic, perturbational, or therapeutic
  evidence to select locus-level associations.

The resulting CpG–gene relationships constitute secondary locus-level molecular
context for the frozen consensus programs.

## Expected outputs

Notebook 451 will produce reproducible locus-level characterization under:

`data/processed/secondary_characterization/`

The analysis will preserve:

- the deterministic strong-loading CpG universe and CpG–gene annotation
  eligibility;
- the complete frozen primary CpG–gene association family;
- lineage-aware recurrence characterization;
- prespecified sensitivity evidence;
- promoter/regulatory-context annotation;
- consensus-program and methylation-component context; and
- explicit negative, heterogeneous, and technically limited results.

Only downstream-consumable artifacts justified by the completed analysis will be
persisted and registered. Intermediate diagnostic objects remain reproducible
within the notebook unless separate persistence is required for downstream
provenance or reuse.

```python
# =============================================================================
# Define frozen notebook-451 methylation-component context
# =============================================================================

methylation_component_context = (
    consensus_tumor_arm_context[
        [
            "consensus_program_id",
            "candidate_pair",
            "rna_component",
            "methylation_component",
            "final_audit_status",
            "scientific_priority",
            "primary_limitation",
            "consensus_handling",
        ]
    ]
    .drop_duplicates()
    .sort_values(
        ["consensus_program_id", "methylation_component", "candidate_pair"]
    )
    .reset_index(drop=True)
)

relevant_methylation_components = tuple(
    sorted(
        methylation_component_context[
            "methylation_component"
        ].dropna().unique()
    )
)

missing_loading_components = sorted(
    set(relevant_methylation_components)
    - set(methylation_probe_loadings.columns)
)

print(
    "Consensus programs represented:",
    methylation_component_context["consensus_program_id"].nunique(),
)
print(
    "Tumor methylation arms represented:",
    methylation_component_context.shape[0],
)
print(
    "Unique methylation components:",
    relevant_methylation_components,
)
print(
    "All relevant components present in frozen loadings:",
    not missing_loading_components,
)

methylation_component_context
```

```python
# =============================================================================
# Construct strong-loading CpG membership universe
# =============================================================================

LOADING_TAIL_SIZE = 50

strong_loading_memberships = []

for component in relevant_methylation_components:
    component_loadings = methylation_probe_loadings[
        ["model_matrix_row_index", "probe_id", component]
    ].rename(columns={component: "loading"})

    negative_tail = (
        component_loadings
        .sort_values(
            ["loading", "probe_id"],
            ascending=[True, True],
        )
        .head(LOADING_TAIL_SIZE)
        .assign(
            methylation_component=component,
            loading_tail="negative",
        )
    )

    positive_tail = (
        component_loadings
        .sort_values(
            ["loading", "probe_id"],
            ascending=[False, True],
        )
        .head(LOADING_TAIL_SIZE)
        .assign(
            methylation_component=component,
            loading_tail="positive",
        )
    )

    strong_loading_memberships.extend(
        [negative_tail, positive_tail]
    )

strong_loading_cpg_memberships = (
    pd.concat(
        strong_loading_memberships,
        ignore_index=True,
    )
    [
        [
            "probe_id",
            "model_matrix_row_index",
            "methylation_component",
            "loading",
            "loading_tail",
        ]
    ]
    .sort_values(
        ["methylation_component", "loading_tail", "loading"],
        ascending=[True, True, True],
    )
    .reset_index(drop=True)
)

expected_memberships = (
    len(relevant_methylation_components)
    * 2
    * LOADING_TAIL_SIZE
)

print("Expected component-tail memberships:", expected_memberships)
print(
    "Observed component-tail memberships:",
    strong_loading_cpg_memberships.shape[0],
)
print(
    "Unique strong-loading CpGs:",
    strong_loading_cpg_memberships["probe_id"].nunique(),
)
print(
    "Duplicate membership rows:",
    strong_loading_cpg_memberships.duplicated(
        ["probe_id", "methylation_component", "loading_tail"]
    ).sum(),
)

strong_loading_cpg_memberships.groupby(
    ["methylation_component", "loading_tail"]
).size()
```

```python
# =============================================================================
# Construct unique strong-loading CpG locus universe
# =============================================================================

probe_index_counts = (
    strong_loading_cpg_memberships
    .groupby("probe_id")["model_matrix_row_index"]
    .nunique()
)

if not probe_index_counts.eq(1).all():
    raise ValueError(
        "At least one strong-loading CpG maps to multiple model matrix rows."
    )

strong_loading_cpg_universe = (
    strong_loading_cpg_memberships
    .groupby(
        ["probe_id", "model_matrix_row_index"],
        as_index=False,
    )
    .agg(
        n_methylation_components=(
            "methylation_component",
            "nunique",
        ),
        methylation_components=(
            "methylation_component",
            lambda values: "|".join(sorted(set(values))),
        ),
        component_tail_memberships=(
            "methylation_component",
            lambda values: "|".join(
                sorted(set(values))
            ),
        ),
    )
)

component_tail_context = (
    strong_loading_cpg_memberships
    .assign(
        component_tail=lambda frame: (
            frame["methylation_component"]
            + ":"
            + frame["loading_tail"]
        )
    )
    .groupby("probe_id")["component_tail"]
    .agg(lambda values: "|".join(sorted(set(values))))
)

strong_loading_cpg_universe[
    "component_tail_memberships"
] = strong_loading_cpg_universe["probe_id"].map(
    component_tail_context
)

strong_loading_cpg_universe = (
    strong_loading_cpg_universe
    .sort_values("probe_id")
    .reset_index(drop=True)
)

print(
    "Unique strong-loading CpGs:",
    strong_loading_cpg_universe.shape[0],
)
print(
    "CpGs represented in >1 methylation component:",
    int(
        strong_loading_cpg_universe[
            "n_methylation_components"
        ].gt(1).sum()
    ),
)
print(
    "Maximum component memberships for one CpG:",
    int(
        strong_loading_cpg_universe[
            "n_methylation_components"
        ].max()
    ),
)

strong_loading_cpg_universe[
    "n_methylation_components"
].value_counts().sort_index()
```

```python
# =============================================================================
# Map strong-loading CpGs to frozen HM450 matrix rows
# =============================================================================

hm450_probe_mapping = (
    probe_mapping.loc[
        probe_mapping["representation"].eq("hm450"),
        [
            "probe_id",
            "matrix_row_index",
            "probe_missing_beta_fraction",
            "probe_qc_eligible",
            "hm450_missing_beta_fraction",
            "hm450_probe_qc_eligible",
        ],
    ]
    .rename(
        columns={
            "matrix_row_index": "hm450_matrix_row_index",
        }
    )
)

hm450_strong_probe_mapping = (
    strong_loading_cpg_universe
    .merge(
        hm450_probe_mapping,
        on="probe_id",
        how="left",
        validate="one_to_one",
    )
    .sort_values("probe_id")
    .reset_index(drop=True)
)

if hm450_strong_probe_mapping["hm450_matrix_row_index"].isna().any():
    raise ValueError(
        "At least one strong-loading CpG lacks an HM450 matrix-row mapping."
    )

print(
    "Strong-loading CpGs:",
    hm450_strong_probe_mapping.shape[0],
)
print(
    "Mapped to HM450 rows:",
    hm450_strong_probe_mapping[
        "hm450_matrix_row_index"
    ].notna().sum(),
)
print(
    "HM450 QC-eligible:",
    hm450_strong_probe_mapping[
        "hm450_probe_qc_eligible"
    ].fillna(False).sum(),
)
print(
    "Unique HM450 matrix rows:",
    hm450_strong_probe_mapping[
        "hm450_matrix_row_index"
    ].nunique(),
)
```

### **CpG biological-annotation checkpoint**

The strong-loading CpG universe and the frozen RNA expression gene axis have now
been defined independently of methylation–expression association results.

Before constructing CpG–gene pairs, notebook 451 requires one authoritative,
versioned biological annotation resource for HM450 probes.

The annotation resource must provide, where available:

- CpG probe identity;
- genomic coordinates;
- genome build;
- gene association;
- promoter or other regulatory-context annotation; and
- sufficient provenance to reproduce the mapping deterministically.

The annotation authority must be frozen before any methylation–expression
association is calculated.

Primary CpG-to-gene assignment will follow the prespecified identifier policy:

1. exact Ensembl base-ID matching when the annotation resource provides a
   compatible stable Ensembl gene identifier;
2. otherwise exact gene-symbol matching only when the symbol is unambiguous in
   the frozen RNA expression matrix.

No fuzzy matching, nearest-gene rescue, manual alias substitution, or
result-driven reassignment is permitted in the primary analysis.

The frozen RNA expression representation contains 18,123 genes with unique
Ensembl base identifiers. Gene-symbol mapping is unambiguous for 18,121 genes.
`POLR2J3` is represented by two distinct Ensembl genes and therefore cannot be
assigned through symbol-only matching.

CpGs mapping to multiple defensible genes will retain every valid CpG–gene
relationship as separate candidate pairs. Repeated transcript-level annotations
of the same CpG–gene relationship will later be collapsed to one unique
inferential CpG–gene unit while preserving annotation metadata.

No methylation–expression inference proceeds until the annotation resource,
version, genome build, provenance, and local frozen representation have been
defined and audited.

### **Frozen CpG annotation authority**

Notebook 451 uses a single prespecified biological annotation authority for
HM450 CpG probes:

- annotation framework: `SeSAMe / sesameData`;
- methylation platform: `HM450`;
- genome build: `hg38`.

This decision is frozen before inspection of any locus-level
methylation–expression association.

The annotation resource will be used only to establish deterministic biological
context for the frozen strong-loading CpG universe, including, where available:

- genomic coordinates;
- CpG-to-gene relationships; and
- promoter or other regulatory-context annotation.

The selected annotation authority does not replace or modify the frozen
technical probe mapping produced in Phase 2. The Phase 2 probe mapping remains
authoritative for matrix-row identity, QC eligibility, platform membership, and
missingness.

Notebook 451 will therefore preserve two distinct annotation layers:

1. **technical probe identity and matrix mapping** inherited from Phase 2; and
2. **biological CpG annotation** derived from the frozen SeSAMe HM450 hg38
   resource.

No alternative annotation source may be substituted after association results
are inspected because it yields more favorable CpG-to-gene mappings or
regulatory interpretation.

The exact SeSAMe / sesameData version, resource identity, retrieval procedure,

```python
# =============================================================================
# Extract HM450 beta values for RNA-resolved CpGs
# =============================================================================

resolved_cpg_hm450_index = (
    rna_resolved_cpg_gene_pairs[["probe_id"]]
    .drop_duplicates()
    .merge(
        hm450_strong_probe_mapping[
            ["probe_id", "hm450_matrix_row_index"]
        ],
        on="probe_id",
        how="left",
        validate="one_to_one",
    )
    .sort_values("probe_id")
    .reset_index(drop=True)
)

resolved_hm450_row_indices = resolved_cpg_hm450_index[
    "hm450_matrix_row_index"
].to_numpy(dtype=np.int64)

resolved_hm450_betas = np.empty(
    (
        len(resolved_hm450_row_indices),
        hm450_beta_values.shape[1],
    ),
    dtype=hm450_beta_values.dtype,
)

column_chunk_size = 64

with open(hm450_beta_values.filename, "rb") as handle:
    handle.seek(hm450_beta_values.offset)

    for column_start in range(
        0,
        hm450_beta_values.shape[1],
        column_chunk_size,
    ):
        column_stop = min(
            column_start + column_chunk_size,
            hm450_beta_values.shape[1],
        )
        n_columns = column_stop - column_start

        beta_block = np.fromfile(
            handle,
            dtype=hm450_beta_values.dtype,
            count=hm450_beta_values.shape[0] * n_columns,
        ).reshape(
            hm450_beta_values.shape[0],
            n_columns,
            order="F",
        )

        resolved_hm450_betas[:, column_start:column_stop] = (
            beta_block[resolved_hm450_row_indices]
        )

print("Extracted HM450 beta matrix:", resolved_hm450_betas.shape)
```

```python
# =============================================================================
# Inspect beta-value boundaries before M-value transformation
# =============================================================================

finite_beta_mask = np.isfinite(resolved_hm450_betas)

print(
    "Missing beta values:",
    int((~finite_beta_mask).sum()),
)
print(
    "Finite beta values equal to 0:",
    int((resolved_hm450_betas[finite_beta_mask] == 0).sum()),
)
print(
    "Finite beta values equal to 1:",
    int((resolved_hm450_betas[finite_beta_mask] == 1).sum()),
)
print(
    "Finite beta range:",
    (
        float(resolved_hm450_betas[finite_beta_mask].min()),
        float(resolved_hm450_betas[finite_beta_mask].max()),
    ),
)
```

```python
# =============================================================================
# Transform HM450 beta values to M-values
# =============================================================================

resolved_hm450_m_values = np.log2(
    resolved_hm450_betas
    / (1.0 - resolved_hm450_betas)
)

print(
    "Resolved HM450 M-value matrix:",
    resolved_hm450_m_values.shape,
)
print(
    "Missing M-values:",
    int(np.isnan(resolved_hm450_m_values).sum()),
)
```

```python
# =============================================================================
# Align frozen RNA sample axis to HM450 tumors
# =============================================================================

hm450_rna_column_indices = hm450_sample_index[
    "final_sample_column_index"
].to_numpy(dtype=np.int64)

if not np.array_equal(
    hm450_sample_index["hm450_matrix_column_index"].to_numpy(),
    np.arange(len(hm450_sample_index)),
):
    raise ValueError(
        "HM450 sample order is not aligned to the frozen HM450 matrix."
    )

print(
    "HM450 tumors with RNA correspondence:",
    len(hm450_rna_column_indices),
)
print(
    "RNA column index range:",
    (
        int(hm450_rna_column_indices.min()),
        int(hm450_rna_column_indices.max()),
    ),
)
```

```python
# =============================================================================
# Construct RNA-resolved gene-axis index
# =============================================================================

resolved_rna_gene_index = (
    rna_resolved_cpg_gene_pairs[
        [
            "rna_gene_id_base",
            "rna_gene_name",
            "rna_logcpm_row_index",
        ]
    ]
    .drop_duplicates()
    .sort_values("rna_logcpm_row_index")
    .reset_index(drop=True)
)

if resolved_rna_gene_index["rna_gene_id_base"].duplicated().any():
    raise ValueError(
        "A resolved RNA gene maps to more than one expression row."
    )

print(
    "RNA-resolved genes:",
    len(resolved_rna_gene_index),
)
print(
    "RNA row index range:",
    (
        int(resolved_rna_gene_index["rna_logcpm_row_index"].min()),
        int(resolved_rna_gene_index["rna_logcpm_row_index"].max()),
    ),
)
```

```python
# =============================================================================
# Extract RNA expression for the HM450 cohort
# =============================================================================

resolved_rna_row_indices = resolved_rna_gene_index[
    "rna_logcpm_row_index"
].to_numpy(dtype=np.int64)

with h5py.File(rna_expression_path, "r") as handle:
    resolved_rna_logcpm = np.asarray(
        handle["logcpm"][resolved_rna_row_indices, :]
    )[:, hm450_rna_column_indices]

print(
    "Resolved RNA logCPM matrix:",
    resolved_rna_logcpm.shape,
)
print(
    "Missing RNA values:",
    int(np.isnan(resolved_rna_logcpm).sum()),
)
```

```python
# =============================================================================
# Link CpG-gene candidate pairs to extracted matrix rows
# =============================================================================

cpg_matrix_index = resolved_cpg_hm450_index[["probe_id"]].copy()
cpg_matrix_index["resolved_hm450_row_index"] = np.arange(
    len(cpg_matrix_index),
    dtype=np.int64,
)

rna_matrix_index = resolved_rna_gene_index[["rna_gene_id_base"]].copy()
rna_matrix_index["resolved_rna_row_index"] = np.arange(
    len(rna_matrix_index),
    dtype=np.int64,
)

cpg_gene_support_candidates = (
    rna_resolved_cpg_gene_pairs
    .merge(
        cpg_matrix_index,
        on="probe_id",
        how="left",
        validate="many_to_one",
    )
    .merge(
        rna_matrix_index,
        on="rna_gene_id_base",
        how="left",
        validate="many_to_one",
    )
)

if cpg_gene_support_candidates[
    ["resolved_hm450_row_index", "resolved_rna_row_index"]
].isna().any().any():
    raise ValueError(
        "At least one CpG-gene pair is not represented in the extracted matrices."
    )

print(
    "CpG-gene support candidates:",
    len(cpg_gene_support_candidates),
)
print(
    "Unique CpGs:",
    cpg_gene_support_candidates["probe_id"].nunique(),
)
print(
    "Unique RNA genes:",
    cpg_gene_support_candidates["rna_gene_id_base"].nunique(),
)
```

```python
# =============================================================================
# Evaluate project-level support for CpG-gene candidate pairs
# =============================================================================

project_column_indices = {
    project_id: group.index.to_numpy(dtype=np.int64)
    for project_id, group in hm450_sample_index.groupby(
        "project_id",
        sort=True,
    )
}

project_support_records = []

for pair in cpg_gene_support_candidates.itertuples(index=False):
    methylation_values = resolved_hm450_m_values[
        pair.resolved_hm450_row_index
    ]
    expression_values = resolved_rna_logcpm[
        pair.resolved_rna_row_index
    ]

    for project_id, column_indices in project_column_indices.items():
        project_methylation = methylation_values[column_indices]
        project_expression = expression_values[column_indices]

        complete_mask = (
            np.isfinite(project_methylation)
            & np.isfinite(project_expression)
        )
        n_complete = int(complete_mask.sum())

        supported = False

        if n_complete >= 20:
            complete_methylation = project_methylation[complete_mask]
            complete_expression = project_expression[complete_mask]

            supported = (
                np.ptp(complete_methylation) > 0
                and np.ptp(complete_expression) > 0
            )

        project_support_records.append(
            {
                "probe_id": pair.probe_id,
                "rna_gene_id_base": pair.rna_gene_id_base,
                "project_id": project_id,
                "n_complete": n_complete,
                "supported": supported,
            }
        )

cpg_gene_project_support = pd.DataFrame(project_support_records)

print(
    "CpG-gene-project combinations:",
    len(cpg_gene_project_support),
)
print(
    "Supported combinations:",
    int(cpg_gene_project_support["supported"].sum()),
)
```

```python
# =============================================================================
# Define the primary CpG-gene inferential family
# =============================================================================

cpg_gene_support_summary = (
    cpg_gene_project_support
    .groupby(
        ["probe_id", "rna_gene_id_base"],
        as_index=False,
    )
    .agg(
        n_supported_projects=("supported", "sum"),
    )
)

primary_cpg_gene_family = (
    cpg_gene_support_candidates
    .merge(
        cpg_gene_support_summary,
        on=["probe_id", "rna_gene_id_base"],
        how="left",
        validate="one_to_one",
    )
    .loc[lambda df: df["n_supported_projects"] >= 3]
    .copy()
    .sort_values(["probe_id", "rna_gene_id_base"])
    .reset_index(drop=True)
)

print(
    "Candidate CpG-gene pairs:",
    len(cpg_gene_support_candidates),
)
print(
    "Primary inferential pairs:",
    len(primary_cpg_gene_family),
)
print(
    "Excluded for insufficient project support:",
    len(cpg_gene_support_candidates) - len(primary_cpg_gene_family),
)
print(
    "Supported-project range:",
    (
        int(primary_cpg_gene_family["n_supported_projects"].min()),
        int(primary_cpg_gene_family["n_supported_projects"].max()),
    ),
)
```

```python
# =============================================================================
# Characterize unsupported project-level pair combinations
# =============================================================================

unsupported_project_summary = (
    cpg_gene_project_support.loc[
        ~cpg_gene_project_support["supported"]
    ]
    .groupby("project_id", as_index=False)
    .agg(
        n_unsupported_pairs=("probe_id", "size"),
        min_complete_cases=("n_complete", "min"),
        max_complete_cases=("n_complete", "max"),
    )
    .sort_values(
        ["n_unsupported_pairs", "project_id"],
        ascending=[False, True],
    )
    .reset_index(drop=True)
)

unsupported_project_summary
```

```python
# =============================================================================
# Construct the primary supported-project sample index
# =============================================================================

primary_supported_projects = sorted(
    cpg_gene_project_support.loc[
        cpg_gene_project_support["supported"],
        "project_id",
    ].unique()
)

primary_sample_mask = hm450_sample_index[
    "project_id"
].isin(primary_supported_projects).to_numpy()

primary_sample_index = (
    hm450_sample_index.loc[primary_sample_mask]
    .reset_index(drop=True)
)

print(
    "Primary supported projects:",
    len(primary_supported_projects),
)
print(
    "Primary HM450 tumors:",
    len(primary_sample_index),
)
print(
    "Excluded HM450 tumors:",
    len(hm450_sample_index) - len(primary_sample_index),
)
```

```python
# =============================================================================
# Construct project fixed-effect design
# =============================================================================

primary_project = pd.Categorical(
    primary_sample_index["project_id"],
    categories=primary_supported_projects,
    ordered=True,
)

project_design = pd.get_dummies(
    primary_project,
    drop_first=True,
    dtype=float,
)

project_design.insert(0, "intercept", 1.0)

primary_column_indices = np.flatnonzero(primary_sample_mask)

print(
    "Primary design samples:",
    len(project_design),
)
print(
    "Project fixed-effect columns:",
    project_design.shape[1],
)
print(
    "Reference project:",
    primary_supported_projects[0],
)
```

```python
# =============================================================================
# Validate the primary model on one CpG-gene pair
# =============================================================================

validation_pair = primary_cpg_gene_family.iloc[0]

methylation_values = resolved_hm450_m_values[
    int(validation_pair["resolved_hm450_row_index"]),
    primary_column_indices,
]

expression_values = resolved_rna_logcpm[
    int(validation_pair["resolved_rna_row_index"]),
    primary_column_indices,
]

complete_mask = (
    np.isfinite(methylation_values)
    & np.isfinite(expression_values)
)

validation_design = project_design.loc[complete_mask].copy()
validation_design["cpg_methylation"] = methylation_values[complete_mask]

validation_model = sm.OLS(
    expression_values[complete_mask],
    validation_design,
).fit(cov_type="HC3")

confidence_interval = validation_model.conf_int().loc[
    "cpg_methylation"
]

print(
    "CpG-gene pair:",
    validation_pair["probe_id"],
    validation_pair["rna_gene_id_base"],
)
print("Complete cases:", int(complete_mask.sum()))
print(
    "Methylation coefficient:",
    float(validation_model.params["cpg_methylation"]),
)
print(
    "HC3 standard error:",
    float(validation_model.bse["cpg_methylation"]),
)
print(
    "95% CI:",
    (
        float(confidence_interval.iloc[0]),
        float(confidence_interval.iloc[1]),
    ),
)
print(
    "Test statistic:",
    float(validation_model.tvalues["cpg_methylation"]),
)
print(
    "P-value:",
    float(validation_model.pvalues["cpg_methylation"]),
)
```

```python
# =============================================================================
# Fit primary CpG-gene association models
# =============================================================================

primary_model_records = []

for pair in primary_cpg_gene_family.itertuples(index=False):
    methylation_values = resolved_hm450_m_values[
        pair.resolved_hm450_row_index,
        primary_column_indices,
    ]
    expression_values = resolved_rna_logcpm[
        pair.resolved_rna_row_index,
        primary_column_indices,
    ]

    complete_mask = (
        np.isfinite(methylation_values)
        & np.isfinite(expression_values)
    )

    model_design = project_design.loc[complete_mask].copy()
    model_design["cpg_methylation"] = methylation_values[complete_mask]

    model = sm.OLS(
        expression_values[complete_mask],
        model_design,
    ).fit(cov_type="HC3")

    confidence_interval = model.conf_int().loc["cpg_methylation"]

    primary_model_records.append(
        {
            "probe_id": pair.probe_id,
            "rna_gene_id_base": pair.rna_gene_id_base,
            "n_complete": int(complete_mask.sum()),
            "n_supported_projects": int(pair.n_supported_projects),
            "methylation_coefficient": float(
                model.params["cpg_methylation"]
            ),
            "hc3_standard_error": float(
                model.bse["cpg_methylation"]
            ),
            "ci_95_lower": float(confidence_interval.iloc[0]),
            "ci_95_upper": float(confidence_interval.iloc[1]),
            "test_statistic": float(
                model.tvalues["cpg_methylation"]
            ),
            "p_value": float(
                model.pvalues["cpg_methylation"]
            ),
        }
    )

primary_model_results = pd.DataFrame(primary_model_records)

print(
    "Primary models fitted:",
    len(primary_model_results),
)
print(
    "Complete-case range:",
    (
        int(primary_model_results["n_complete"].min()),
        int(primary_model_results["n_complete"].max()),
    ),
)
print(
    "Nominal p-value range:",
    (
        float(primary_model_results["p_value"].min()),
        float(primary_model_results["p_value"].max()),
    ),
)
```

```python
# =============================================================================
# Apply global Benjamini-Hochberg correction
# =============================================================================

fdr_reject, q_values, _, _ = sm.stats.multipletests(
    primary_model_results["p_value"].to_numpy(),
    alpha=0.05,
    method="fdr_bh",
)

primary_model_results["q_value"] = q_values
primary_model_results["primary_fdr_supported"] = fdr_reject

print(
    "Primary hypothesis tests:",
    len(primary_model_results),
)
print(
    "FDR-supported associations:",
    int(primary_model_results["primary_fdr_supported"].sum()),
)
print(
    "Q-value range:",
    (
        float(primary_model_results["q_value"].min()),
        float(primary_model_results["q_value"].max()),
    ),
)
print(
    "Numerically zero p-values:",
    int(primary_model_results["p_value"].eq(0.0).sum()),
)
```

```python
# =============================================================================
# Define primary FDR-supported associations
# =============================================================================

primary_fdr_associations = (
    primary_cpg_gene_family
    .merge(
        primary_model_results,
        on=[
            "probe_id",
            "rna_gene_id_base",
            "n_supported_projects",
        ],
        how="inner",
        validate="one_to_one",
    )
    .loc[lambda df: df["primary_fdr_supported"]]
    .copy()
    .sort_values(["probe_id", "rna_gene_id_base"])
    .reset_index(drop=True)
)

print(
    "Primary FDR-supported associations:",
    len(primary_fdr_associations),
)
print(
    "Inverse associations:",
    int(
        primary_fdr_associations[
            "methylation_coefficient"
        ].lt(0).sum()
    ),
)
print(
    "Positive associations:",
    int(
        primary_fdr_associations[
            "methylation_coefficient"
        ].gt(0).sum()
    ),
)
```

```python
# =============================================================================
# Estimate within-project CpG-gene associations
# =============================================================================

supported_projects_by_pair = (
    cpg_gene_project_support.loc[
        cpg_gene_project_support["supported"]
    ]
    .groupby(["probe_id", "rna_gene_id_base"])["project_id"]
    .apply(list)
    .to_dict()
)

within_project_records = []

for pair in primary_fdr_associations.itertuples(index=False):
    methylation_values = resolved_hm450_m_values[
        int(pair.resolved_hm450_row_index)
    ]
    expression_values = resolved_rna_logcpm[
        int(pair.resolved_rna_row_index)
    ]

    pair_key = (
        pair.probe_id,
        pair.rna_gene_id_base,
    )

    for project_id in supported_projects_by_pair[pair_key]:
        column_indices = project_column_indices[project_id]

        project_methylation = methylation_values[column_indices]
        project_expression = expression_values[column_indices]

        complete_mask = (
            np.isfinite(project_methylation)
            & np.isfinite(project_expression)
        )

        model_design = sm.add_constant(
            project_methylation[complete_mask],
            has_constant="add",
        )

        model = sm.OLS(
            project_expression[complete_mask],
            model_design,
        ).fit()

        within_project_records.append(
            {
                "probe_id": pair.probe_id,
                "rna_gene_id_base": pair.rna_gene_id_base,
                "project_id": project_id,
                "n_complete": int(complete_mask.sum()),
                "methylation_coefficient": float(model.params[1]),
            }
        )

within_project_results = pd.DataFrame(within_project_records)

project_counts = (
    within_project_results
    .groupby(["probe_id", "rna_gene_id_base"])
    .size()
)

print(
    "Within-project associations:",
    len(within_project_results),
)
print(
    "Projects per FDR-supported association:",
    (
        int(project_counts.min()),
        int(project_counts.max()),
    ),
)
```

```python
# =============================================================================
# Summarize cross-project directional concordance
# =============================================================================

pooled_direction = (
    primary_fdr_associations[
        [
            "probe_id",
            "rna_gene_id_base",
            "methylation_coefficient",
        ]
    ]
    .rename(
        columns={
            "methylation_coefficient": "pooled_methylation_coefficient"
        }
    )
)

within_project_direction = (
    within_project_results
    .merge(
        pooled_direction,
        on=["probe_id", "rna_gene_id_base"],
        how="left",
        validate="many_to_one",
    )
)

within_project_direction["same_pooled_direction"] = (
    np.sign(within_project_direction["methylation_coefficient"])
    == np.sign(
        within_project_direction["pooled_methylation_coefficient"]
    )
)

directional_concordance = (
    within_project_direction
    .groupby(
        ["probe_id", "rna_gene_id_base"],
        as_index=False,
    )
    .agg(
        n_supported_projects=("project_id", "size"),
        n_same_direction_projects=("same_pooled_direction", "sum"),
    )
)

directional_concordance["directional_concordance_fraction"] = (
    directional_concordance["n_same_direction_projects"]
    / directional_concordance["n_supported_projects"]
)

directional_concordance["passes_directional_criteria"] = (
    directional_concordance["n_same_direction_projects"].ge(3)
    & directional_concordance[
        "directional_concordance_fraction"
    ].ge(0.70)
)

print(
    "FDR-supported associations:",
    len(directional_concordance),
)
print(
    "Passing directional criteria:",
    int(
        directional_concordance[
            "passes_directional_criteria"
        ].sum()
    ),
)
print(
    "Directional concordance range:",
    (
        float(
            directional_concordance[
                "directional_concordance_fraction"
            ].min()
        ),
        float(
            directional_concordance[
                "directional_concordance_fraction"
            ].max()
        ),
    ),
)
```

```python
# =============================================================================
# Validate fixed-effect residualization against the primary model
# =============================================================================

validation_pair = primary_fdr_associations.iloc[0]

methylation_values = resolved_hm450_m_values[
    int(validation_pair["resolved_hm450_row_index"]),
    primary_column_indices,
]

expression_values = resolved_rna_logcpm[
    int(validation_pair["resolved_rna_row_index"]),
    primary_column_indices,
]

complete_mask = (
    np.isfinite(methylation_values)
    & np.isfinite(expression_values)
)

validation_data = pd.DataFrame(
    {
        "methylation": methylation_values[complete_mask],
        "expression": expression_values[complete_mask],
        "project_id": primary_sample_index.loc[
            complete_mask,
            "project_id",
        ].to_numpy(),
    }
)

validation_data["methylation_centered"] = (
    validation_data["methylation"]
    - validation_data.groupby("project_id")["methylation"].transform("mean")
)

validation_data["expression_centered"] = (
    validation_data["expression"]
    - validation_data.groupby("project_id")["expression"].transform("mean")
)

residualized_coefficient = (
    np.dot(
        validation_data["methylation_centered"],
        validation_data["expression_centered"],
    )
    / np.dot(
        validation_data["methylation_centered"],
        validation_data["methylation_centered"],
    )
)

primary_coefficient = validation_pair["methylation_coefficient"]

print(
    "Primary-model coefficient:",
    float(primary_coefficient),
)
print(
    "Residualized coefficient:",
    float(residualized_coefficient),
)
print(
    "Absolute difference:",
    float(abs(primary_coefficient - residualized_coefficient)),
)
```

```python
# =============================================================================
# Perform leave-one-project-out coefficient estimation
# =============================================================================

lopo_records = []

for pair in primary_fdr_associations.itertuples(index=False):
    methylation_values = resolved_hm450_m_values[
        int(pair.resolved_hm450_row_index)
    ]
    expression_values = resolved_rna_logcpm[
        int(pair.resolved_rna_row_index)
    ]

    pair_key = (
        pair.probe_id,
        pair.rna_gene_id_base,
    )

    project_sums = []

    for project_id in supported_projects_by_pair[pair_key]:
        column_indices = project_column_indices[project_id]

        project_methylation = methylation_values[column_indices]
        project_expression = expression_values[column_indices]

        complete_mask = (
            np.isfinite(project_methylation)
            & np.isfinite(project_expression)
        )

        x = project_methylation[complete_mask]
        y = project_expression[complete_mask]

        x_centered = x - x.mean()
        y_centered = y - y.mean()

        project_sums.append(
            {
                "project_id": project_id,
                "n_complete": int(complete_mask.sum()),
                "sxx": float(np.dot(x_centered, x_centered)),
                "sxy": float(np.dot(x_centered, y_centered)),
            }
        )

    total_sxx = sum(item["sxx"] for item in project_sums)
    total_sxy = sum(item["sxy"] for item in project_sums)
    total_n = sum(item["n_complete"] for item in project_sums)

    for omitted in project_sums:
        lopo_coefficient = (
            (total_sxy - omitted["sxy"])
            / (total_sxx - omitted["sxx"])
        )

        lopo_records.append(
            {
                "probe_id": pair.probe_id,
                "rna_gene_id_base": pair.rna_gene_id_base,
                "omitted_project_id": omitted["project_id"],
                "n_complete": total_n - omitted["n_complete"],
                "lopo_methylation_coefficient": lopo_coefficient,
                "sign_reversal": (
                    lopo_coefficient
                    * pair.methylation_coefficient
                    < 0
                ),
            }
        )

lopo_results = pd.DataFrame(lopo_records)

print(
    "LOPO refits:",
    len(lopo_results),
)
print(
    "Associations with >=1 sign reversal:",
    int(
        lopo_results
        .groupby(["probe_id", "rna_gene_id_base"])["sign_reversal"]
        .any()
        .sum()
    ),
)
print(
    "Minimum absolute LOPO coefficient:",
    float(
        lopo_results["lopo_methylation_coefficient"]
        .abs()
        .min()
    ),
)
```

```python
# =============================================================================
# Classify cross-project recurrent associations
# =============================================================================

lopo_summary = (
    lopo_results
    .groupby(
        ["probe_id", "rna_gene_id_base"],
        as_index=False,
    )
    .agg(
        any_lopo_sign_reversal=("sign_reversal", "any"),
    )
)

recurrence_summary = (
    primary_fdr_associations[
        [
            "probe_id",
            "rna_gene_id_base",
            "n_supported_projects",
            "methylation_coefficient",
            "q_value",
        ]
    ]
    .merge(
        directional_concordance,
        on=["probe_id", "rna_gene_id_base"],
        how="left",
        validate="one_to_one",
        suffixes=("", "_directional"),
    )
    .merge(
        lopo_summary,
        on=["probe_id", "rna_gene_id_base"],
        how="left",
        validate="one_to_one",
    )
)

recurrence_summary["cross_project_recurrent"] = (
    recurrence_summary["q_value"].lt(0.05)
    & recurrence_summary["n_supported_projects"].ge(3)
    & recurrence_summary["n_same_direction_projects"].ge(3)
    & recurrence_summary["directional_concordance_fraction"].ge(0.70)
    & ~recurrence_summary["any_lopo_sign_reversal"]
)

print(
    "FDR-supported associations:",
    len(recurrence_summary),
)
print(
    "Cross-project recurrent associations:",
    int(recurrence_summary["cross_project_recurrent"].sum()),
)
print(
    "Directional-pass associations with LOPO reversal:",
    int(
        (
            recurrence_summary["passes_directional_criteria"]
            & recurrence_summary["any_lopo_sign_reversal"]
        ).sum()
    ),
)
print(
    "Non-recurrent associations:",
    int((~recurrence_summary["cross_project_recurrent"]).sum()),
)
```

```python
# =============================================================================
# Consolidate cross-project recurrence with primary associations
# =============================================================================

recurrence_metrics = recurrence_summary[
    [
        "probe_id",
        "rna_gene_id_base",
        "n_same_direction_projects",
        "directional_concordance_fraction",
        "any_lopo_sign_reversal",
        "cross_project_recurrent",
    ]
]

primary_fdr_characterization = (
    primary_fdr_associations
    .merge(
        recurrence_metrics,
        on=["probe_id", "rna_gene_id_base"],
        how="left",
        validate="one_to_one",
    )
)

primary_recurrent_associations = (
    primary_fdr_characterization
    .loc[lambda df: df["cross_project_recurrent"]]
    .copy()
    .sort_values(["probe_id", "rna_gene_id_base"])
    .reset_index(drop=True)
)

print(
    "Cross-project recurrent associations:",
    len(primary_recurrent_associations),
)
print(
    "Recurrent inverse associations:",
    int(
        primary_recurrent_associations[
            "methylation_coefficient"
        ].lt(0).sum()
    ),
)
print(
    "Recurrent positive associations:",
    int(
        primary_recurrent_associations[
            "methylation_coefficient"
        ].gt(0).sum()
    ),
)
```

```python
# =============================================================================
# Fit tumor-purity sensitivity models
# =============================================================================

absolute_purity = primary_sensitivity_covariates[
    "absolute_purity"
].to_numpy(dtype=float)

purity_sensitivity_records = []

for pair in primary_fdr_associations.itertuples(index=False):
    methylation_values = resolved_hm450_m_values[
        int(pair.resolved_hm450_row_index),
        primary_column_indices,
    ]
    expression_values = resolved_rna_logcpm[
        int(pair.resolved_rna_row_index),
        primary_column_indices,
    ]

    complete_mask = (
        np.isfinite(methylation_values)
        & np.isfinite(expression_values)
        & np.isfinite(absolute_purity)
    )

    x = methylation_values[complete_mask]
    y = expression_values[complete_mask]
    purity = absolute_purity[complete_mask]
    projects = primary_sample_index.loc[
        complete_mask,
        "project_id",
    ].to_numpy()

    n_supported_projects = 0

    for project_id in np.unique(projects):
        project_mask = projects == project_id

        if (
            project_mask.sum() >= 20
            and np.ptp(x[project_mask]) > 0
            and np.ptp(y[project_mask]) > 0
        ):
            n_supported_projects += 1

    model_design = pd.get_dummies(
        pd.Categorical(projects),
        drop_first=True,
        dtype=float,
    )
    model_design.insert(0, "intercept", 1.0)
    model_design["cpg_methylation"] = x
    model_design["absolute_purity"] = purity

    model = sm.OLS(
        y,
        model_design,
    ).fit(cov_type="HC3")

    confidence_interval = model.conf_int().loc[
        "cpg_methylation"
    ]

    purity_coefficient = float(
        model.params["cpg_methylation"]
    )
    primary_coefficient = float(
        pair.methylation_coefficient
    )

    purity_sensitivity_records.append(
        {
            "probe_id": pair.probe_id,
            "rna_gene_id_base": pair.rna_gene_id_base,
            "n_complete": int(complete_mask.sum()),
            "n_supported_projects": n_supported_projects,
            "methylation_coefficient": purity_coefficient,
            "hc3_standard_error": float(
                model.bse["cpg_methylation"]
            ),
            "ci_95_lower": float(confidence_interval.iloc[0]),
            "ci_95_upper": float(confidence_interval.iloc[1]),
            "direction_preserved": (
                np.sign(purity_coefficient)
                == np.sign(primary_coefficient)
            ),
            "magnitude_ratio_to_primary": (
                abs(purity_coefficient)
                / abs(primary_coefficient)
            ),
        }
    )

purity_sensitivity_results = pd.DataFrame(
    purity_sensitivity_records
)

print(
    "Purity sensitivity models:",
    len(purity_sensitivity_results),
)
print(
    "Complete-case range:",
    (
        int(purity_sensitivity_results["n_complete"].min()),
        int(purity_sensitivity_results["n_complete"].max()),
    ),
)
print(
    "Supported-project range:",
    (
        int(
            purity_sensitivity_results[
                "n_supported_projects"
            ].min()
        ),
        int(
            purity_sensitivity_results[
                "n_supported_projects"
            ].max()
        ),
    ),
)
print(
    "Direction preserved:",
    int(
        purity_sensitivity_results[
            "direction_preserved"
        ].sum()
    ),
)
```

```python
# =============================================================================
# Align HM27 samples to the frozen RNA expression matrix
# =============================================================================

hm27_rna_column_indices = (
    sample_mapping.loc[
        sample_mapping["hm27_matrix_column_index"].notna(),
        [
            "hm27_matrix_column_index",
            "rna_final_matrix_column_index",
        ],
    ]
    .sort_values("hm27_matrix_column_index")
    ["rna_final_matrix_column_index"]
    .to_numpy(dtype=np.int64)
)

print(
    "HM27-to-RNA correspondences:",
    len(hm27_rna_column_indices),
)
print(
    "Unique RNA columns:",
    np.unique(hm27_rna_column_indices).size,
)
print(
    "RNA column-index range:",
    (
        int(hm27_rna_column_indices.min()),
        int(hm27_rna_column_indices.max()),
    ),
)
```

```python
# =============================================================================
# Evaluate lineage-aware HM27 support
# =============================================================================

hm27_cpg_row_lookup = dict(
    zip(
        hm27_covered_cpg_index["probe_id"],
        range(len(hm27_covered_cpg_index)),
    )
)

hm27_gene_row_lookup = dict(
    zip(
        hm27_gene_index["rna_gene_id_base"],
        range(len(hm27_gene_index)),
    )
)

hm27_projects = hm27_sample_index["project_id"].to_numpy()

hm27_support_records = []

for pair in hm27_covered_fdr_associations.itertuples(index=False):
    x = hm27_covered_m_values[
        hm27_cpg_row_lookup[pair.probe_id]
    ]
    y = hm27_rna_logcpm[
        hm27_gene_row_lookup[pair.rna_gene_id_base]
    ]

    complete_mask = np.isfinite(x) & np.isfinite(y)

    supported_projects = []

    for project_id in np.unique(hm27_projects):
        project_mask = (
            complete_mask
            & (hm27_projects == project_id)
        )

        if (
            project_mask.sum() >= 20
            and np.ptp(x[project_mask]) > 0
            and np.ptp(y[project_mask]) > 0
        ):
            supported_projects.append(project_id)

    hm27_support_records.append(
        {
            "probe_id": pair.probe_id,
            "rna_gene_id_base": pair.rna_gene_id_base,
            "n_complete": int(complete_mask.sum()),
            "n_supported_projects": len(supported_projects),
            "hm27_evaluable": len(supported_projects) >= 3,
        }
    )

hm27_support_summary = pd.DataFrame(
    hm27_support_records
)

print(
    "HM27-covered associations:",
    len(hm27_support_summary),
)
print(
    "HM27-evaluable associations:",
    int(hm27_support_summary["hm27_evaluable"].sum()),
)
print(
    "Complete-case range:",
    (
        int(hm27_support_summary["n_complete"].min()),
        int(hm27_support_summary["n_complete"].max()),
    ),
)
print(
    "Supported-project range:",
    (
        int(hm27_support_summary["n_supported_projects"].min()),
        int(hm27_support_summary["n_supported_projects"].max()),
    ),
)
```

```python
# =============================================================================
# Characterize HM27 project-support composition
# =============================================================================

hm27_pair_project_support_records = []

for pair in hm27_covered_fdr_associations.itertuples(index=False):
    x = hm27_covered_m_values[
        hm27_cpg_row_lookup[pair.probe_id]
    ]
    y = hm27_rna_logcpm[
        hm27_gene_row_lookup[pair.rna_gene_id_base]
    ]

    for project_id in np.unique(hm27_projects):
        project_mask = hm27_projects == project_id
        complete_mask = (
            project_mask
            & np.isfinite(x)
            & np.isfinite(y)
        )

        n_complete = int(complete_mask.sum())

        supported = (
            n_complete >= 20
            and np.ptp(x[complete_mask]) > 0
            and np.ptp(y[complete_mask]) > 0
        )

        hm27_pair_project_support_records.append(
            {
                "probe_id": pair.probe_id,
                "rna_gene_id_base": pair.rna_gene_id_base,
                "project_id": project_id,
                "n_complete": n_complete,
                "supported": supported,
            }
        )

hm27_pair_project_support = pd.DataFrame(
    hm27_pair_project_support_records
)

hm27_unsupported_project_summary = (
    hm27_pair_project_support.loc[
        ~hm27_pair_project_support["supported"]
    ]
    .groupby("project_id", as_index=False)
    .agg(
        n_unsupported_pairs=("probe_id", "size"),
        min_complete_cases=("n_complete", "min"),
        max_complete_cases=("n_complete", "max"),
    )
)

print(
    "Distinct supported-project sets:",
    hm27_pair_project_support.loc[
        hm27_pair_project_support["supported"]
    ]
    .groupby(["probe_id", "rna_gene_id_base"])["project_id"]
    .agg(tuple)
    .nunique(),
)

hm27_unsupported_project_summary
```

```python
# =============================================================================
# Fit HM27 platform-sensitivity models
# =============================================================================

hm27_supported_projects = (
    hm27_pair_project_support.loc[
        hm27_pair_project_support["supported"],
        "project_id",
    ]
    .drop_duplicates()
    .sort_values()
    .to_numpy()
)

hm27_supported_sample_mask = np.isin(
    hm27_projects,
    hm27_supported_projects,
)

hm27_sensitivity_records = []

for pair in hm27_covered_fdr_associations.itertuples(index=False):
    x_all = hm27_covered_m_values[
        hm27_cpg_row_lookup[pair.probe_id]
    ]
    y_all = hm27_rna_logcpm[
        hm27_gene_row_lookup[pair.rna_gene_id_base]
    ]

    complete_mask = (
        hm27_supported_sample_mask
        & np.isfinite(x_all)
        & np.isfinite(y_all)
    )

    x = x_all[complete_mask]
    y = y_all[complete_mask]
    projects = hm27_projects[complete_mask]

    model_design = pd.get_dummies(
        pd.Categorical(
            projects,
            categories=hm27_supported_projects,
        ),
        drop_first=True,
        dtype=float,
    )
    model_design.insert(0, "intercept", 1.0)
    model_design["cpg_methylation"] = x

    model = sm.OLS(
        y,
        model_design,
    ).fit(cov_type="HC3")

    confidence_interval = model.conf_int().loc[
        "cpg_methylation"
    ]

    hm27_coefficient = float(
        model.params["cpg_methylation"]
    )
    primary_coefficient = float(
        pair.methylation_coefficient
    )

    hm27_sensitivity_records.append(
        {
            "probe_id": pair.probe_id,
            "rna_gene_id_base": pair.rna_gene_id_base,
            "n_complete": int(complete_mask.sum()),
            "n_supported_projects": len(
                hm27_supported_projects
            ),
            "methylation_coefficient": hm27_coefficient,
            "hc3_standard_error": float(
                model.bse["cpg_methylation"]
            ),
            "ci_95_lower": float(
                confidence_interval.iloc[0]
            ),
            "ci_95_upper": float(
                confidence_interval.iloc[1]
            ),
            "direction_preserved": (
                np.sign(hm27_coefficient)
                == np.sign(primary_coefficient)
            ),
            "magnitude_ratio_to_primary": (
                abs(hm27_coefficient)
                / abs(primary_coefficient)
            ),
        }
    )

hm27_sensitivity_results = pd.DataFrame(
    hm27_sensitivity_records
)

print(
    "HM27 sensitivity models:",
    len(hm27_sensitivity_results),
)
print(
    "Complete-case range:",
    (
        int(hm27_sensitivity_results["n_complete"].min()),
        int(hm27_sensitivity_results["n_complete"].max()),
    ),
)
print(
    "Supported projects:",
    int(
        hm27_sensitivity_results[
            "n_supported_projects"
        ].min()
    ),
)
print(
    "Direction preserved:",
    int(
        hm27_sensitivity_results[
            "direction_preserved"
        ].sum()
    ),
)
```

```python
# =============================================================================
# Consolidate prespecified sensitivity results
# =============================================================================

association_sensitivities = pd.concat(
    [
        purity_sensitivity_results.assign(
            sensitivity_type="purity",
        ),
        proliferation_sensitivity_results.assign(
            sensitivity_type="proliferation",
        ),
        leukocyte_sensitivity_results.assign(
            sensitivity_type="leukocyte_fraction",
        ),
        hm27_sensitivity_results.assign(
            sensitivity_type="hm27_platform",
        ),
    ],
    ignore_index=True,
)

association_sensitivities = association_sensitivities[
    [
        "probe_id",
        "rna_gene_id_base",
        "sensitivity_type",
        "n_complete",
        "n_supported_projects",
        "methylation_coefficient",
        "hc3_standard_error",
        "ci_95_lower",
        "ci_95_upper",
        "direction_preserved",
        "magnitude_ratio_to_primary",
    ]
]

print(
    "Sensitivity records:",
    len(association_sensitivities),
)

print(
    association_sensitivities
    .groupby("sensitivity_type")
    .agg(
        n_associations=("probe_id", "size"),
        direction_preserved=("direction_preserved", "sum"),
    )
)
```

```python
# =============================================================================
# Attach frozen component and consensus-program context without row expansion
# =============================================================================

cpg_component_context = (
    strong_loading_cpg_memberships
    .assign(
        component_loading_context=lambda df: (
            df["methylation_component"].astype(str)
            + "|"
            + df["loading_tail"].astype(str)
            + "|"
            + df["loading"].astype(str)
        )
    )
    .groupby("probe_id", as_index=False)
    .agg(
        methylation_component_memberships=(
            "methylation_component",
            lambda x: ";".join(sorted(set(x))),
        ),
        component_loading_context=(
            "component_loading_context",
            lambda x: ";".join(sorted(set(x))),
        ),
    )
)

cpg_program_context = (
    strong_loading_cpg_memberships[
        ["probe_id", "methylation_component"]
    ]
    .drop_duplicates()
    .merge(
        methylation_component_context[
            [
                "methylation_component",
                "consensus_program_id",
                "candidate_pair",
                "rna_component",
            ]
        ],
        on="methylation_component",
        how="left",
        validate="many_to_many",
    )
    .assign(
        program_context=lambda df: (
            df["consensus_program_id"].astype(str)
            + "|"
            + df["candidate_pair"].astype(str)
            + "|"
            + df["rna_component"].astype(str)
            + "|"
            + df["methylation_component"].astype(str)
        )
    )
    .groupby("probe_id", as_index=False)
    .agg(
        consensus_program_memberships=(
            "consensus_program_id",
            lambda x: ";".join(sorted(set(x))),
        ),
        program_context=(
            "program_context",
            lambda x: ";".join(sorted(set(x))),
        ),
    )
)

primary_locus_evidence = (
    primary_locus_evidence
    .merge(
        cpg_component_context,
        on="probe_id",
        how="left",
        validate="many_to_one",
    )
    .merge(
        cpg_program_context,
        on="probe_id",
        how="left",
        validate="many_to_one",
    )
)

print(
    "Primary evidence rows:",
    len(primary_locus_evidence),
)
print(
    "Rows with component context:",
    int(
        primary_locus_evidence[
            "methylation_component_memberships"
        ].notna().sum()
    ),
)
print(
    "Rows with consensus-program context:",
    int(
        primary_locus_evidence[
            "consensus_program_memberships"
        ].notna().sum()
    ),
)
print(
    "Unique CpG-gene pairs:",
    primary_locus_evidence[
        ["probe_id", "rna_gene_id_base"]
    ].drop_duplicates().shape[0],
)
```

```python
# =============================================================================
# Characterize locus-level evidence by methylation component
# =============================================================================

component_evidence = (
    primary_locus_evidence[
        [
            "probe_id",
            "rna_gene_id_base",
            "primary_fdr_supported",
            "cross_project_recurrent",
            "fdr_supported_promoter_associated_inverse",
        ]
    ]
    .merge(
        strong_loading_cpg_memberships[
            [
                "probe_id",
                "methylation_component",
            ]
        ].drop_duplicates(),
        on="probe_id",
        how="left",
        validate="many_to_many",
    )
    .drop_duplicates(
        [
            "probe_id",
            "rna_gene_id_base",
            "methylation_component",
        ]
    )
)

component_evidence_summary = (
    component_evidence
    .groupby(
        "methylation_component",
        as_index=False,
    )
    .agg(
        n_primary_pairs=("probe_id", "size"),
        n_fdr_supported=("primary_fdr_supported", "sum"),
        n_cross_project_recurrent=("cross_project_recurrent", "sum"),
        n_fdr_promoter_inverse=(
            "fdr_supported_promoter_associated_inverse",
            "sum",
        ),
    )
)

component_evidence_summary
```

```python
# =============================================================================
# Characterize sensitivity stability by methylation component
# =============================================================================

recurrent_component_sensitivity = (
    recurrent_evidence[
        [
            "probe_id",
            "rna_gene_id_base",
            "purity_direction_preserved",
            "proliferation_direction_preserved",
            "leukocyte_direction_preserved",
            "hm27_direction_preserved",
        ]
    ]
    .merge(
        strong_loading_cpg_memberships[
            ["probe_id", "methylation_component"]
        ].drop_duplicates(),
        on="probe_id",
        how="left",
        validate="many_to_many",
    )
    .drop_duplicates(
        [
            "probe_id",
            "rna_gene_id_base",
            "methylation_component",
        ]
    )
)

recurrent_component_sensitivity[
    "leukocyte_applicable_in_component"
] = recurrent_component_sensitivity[
    "methylation_component"
].eq("METH_IC128")

recurrent_component_sensitivity[
    "leukocyte_preserved_in_component"
] = (
    recurrent_component_sensitivity[
        "leukocyte_applicable_in_component"
    ]
    & recurrent_component_sensitivity[
        "leukocyte_direction_preserved"
    ].fillna(False)
)

component_sensitivity_summary = (
    recurrent_component_sensitivity
    .groupby("methylation_component", as_index=False)
    .agg(
        n_recurrent=("probe_id", "size"),
        n_purity_preserved=("purity_direction_preserved", "sum"),
        n_proliferation_preserved=(
            "proliferation_direction_preserved",
            "sum",
        ),
        n_leukocyte_applicable=(
            "leukocyte_applicable_in_component",
            "sum",
        ),
        n_leukocyte_preserved=(
            "leukocyte_preserved_in_component",
            "sum",
        ),
        n_hm27_preserved=("hm27_direction_preserved", "sum"),
    )
)

component_sensitivity_summary
```

## Results summary and interpretation

Notebook 451 evaluated locus-level methylation–expression relationships within
the prespecified strong-loading CpG context of `METH_IC050`, `METH_IC128`, and
`METH_IC169`.

The frozen strong-loading universe contained 279 unique CpGs. All were covered
by the frozen HM450 hg38 annotation; 272 had at least one protein-coding gene
mapping and 7 had no protein-coding gene mapping. Biological annotation yielded
336 unique CpG–gene pairs, of which 311 mapped deterministically to the frozen
RNA-seq feature axis and entered the complete primary inferential family.
Twenty-five annotated CpG–gene mappings remained unresolved in the frozen RNA
representation and were not rescued.

Primary HM450 models adjusted for TCGA project and used HC3 uncertainty.
Benjamini–Hochberg correction was applied globally across the 311 prespecified
CpG–gene tests. Of these, 289 were FDR-supported at q < 0.05.

Among the FDR-supported associations, 223 satisfied the prespecified
cross-project recurrence criteria, combining directional concordance across
supported TCGA projects with leave-one-project-out sign stability. Of the 223
recurrent associations, 195 were inverse and 28 were positive. A total of 192
recurrent associations were both promoter-associated and inverse, representing
relationships compatible with an epigenetic regulatory hypothesis but not
evidence of methylation-mediated transcriptional repression or causality.

Contextual characterization showed that recurrent evidence was represented
across all three frozen methylation components. Recurrent promoter-associated
inverse relationships accounted for 72/78 contextual recurrent associations in
`METH_IC050`, 86/93 in `METH_IC128`, and 48/68 in `METH_IC169`. Because CpGs can
belong to more than one methylation-component or consensus-program context,
these component-specific counts are overlapping contextual annotations and are
not independent inferential families.

Direction was preserved for all 223 recurrent associations in both the tumor
purity and proliferation sensitivity analyses. Within the prespecified
`METH_IC128` cellular-composition sensitivity, 85/93 recurrent associations
preserved direction after adjustment for leukocyte fraction. This does not
remove the upstream cellularity/immune-composition limitation attached to
`METH_IC128`.

HM27 platform sensitivity was substantially less concordant: 118/223 recurrent
HM450 associations preserved direction in the HM27 analysis. HM27 therefore
provides limited cross-platform compatibility rather than independent
biological validation. Differences in platform coverage, sample composition,
and represented TCGA lineages prevent attribution of this discordance to
platform effects alone.

These results characterize recurrent locus-level molecular associations within
a deliberately selected strong-loading methylation context. The high fraction
of FDR-supported associations must not be interpreted as a genome-wide hit
rate, and cross-project recurrence must not be interpreted as biological
universality across cancers.

Notebook 451 does not establish causal epigenetic regulation, a mechanism of
drug resistance, a clinical predictor, a validated biomarker, or a validated
therapeutic target. The resulting CpG–gene relationships constitute secondary
molecular context for the frozen consensus programs and are intended for
downstream integration with independent functional and pharmacogenomic
evidence.
